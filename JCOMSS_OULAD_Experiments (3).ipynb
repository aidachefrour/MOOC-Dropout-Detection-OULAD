```python

# STEP 3: Investigate duplicates and withdrawal dates

svle = datasets["studentVle"].copy()
reg = datasets["studentRegistration"].copy()
info = datasets["studentInfo"].copy()

# A. Analyze repeated interaction keys
keys = [
    "code_module",
    "code_presentation",
    "id_student",
    "id_site",
    "date"
]

grouped = svle.groupby(keys, dropna=False).agg(
    number_of_rows=("sum_click", "size"),
    distinct_click_values=("sum_click", "nunique"),
    total_clicks=("sum_click", "sum")
).reset_index()

print("===== INTERACTION KEY ANALYSIS =====")
print("Total original rows:", len(svle))
print("Unique interaction keys:", len(grouped))
print("Keys appearing more than once:",
      (grouped["number_of_rows"] > 1).sum())
print("Keys with different sum_click values:",
      (grouped["distinct_click_values"] > 1).sum())
print("Maximum rows per interaction key:",
      grouped["number_of_rows"].max())

print("\nExamples of repeated keys with different click values:")
display(
    grouped[
        (grouped["number_of_rows"] > 1) &
        (grouped["distinct_click_values"] > 1)
    ].head(10)
)

# B. Convert '?' into missing values in unregistration date
reg["date_unregistration_clean"] = (
    reg["date_unregistration"]
    .replace("?", pd.NA)
)

reg["date_unregistration_clean"] = pd.to_numeric(
    reg["date_unregistration_clean"],
    errors="coerce"
)

# C. Join registration dates with final results
student_status = info.merge(
    reg[[
        "code_module",
        "code_presentation",
        "id_student",
        "date_registration",
        "date_unregistration_clean"
    ]],
    on=["code_module", "code_presentation", "id_student"],
    how="left",
    validate="one_to_one"
)

print("\n===== FINAL RESULT VS UNREGISTRATION =====")
print(pd.crosstab(
    student_status["final_result"],
    student_status["date_unregistration_clean"].isna(),
    margins=True
))

print("\n===== WITHDRAWN STUDENTS =====")
withdrawn = student_status[
    student_status["final_result"] == "Withdrawn"
]

print("Total withdrawn:", len(withdrawn))
print("Withdrawn with known unregistration date:",
      withdrawn["date_unregistration_clean"].notna().sum())
print("Withdrawn without known unregistration date:",
      withdrawn["date_unregistration_clean"].isna().sum())

print("\nWithdrawal date statistics:")
display(
    withdrawn["date_unregistration_clean"].describe()
)

print("\nFirst 10 withdrawal records:")
display(
    withdrawn[[
        "code_module",
        "code_presentation",
        "id_student",
        "date_registration",
        "date_unregistration_clean"
    ]].head(10)
)
```

```python

# STEP 4: Define the day-28 prediction cohort

import pandas as pd

PREDICTION_DAY = 28

info = datasets["studentInfo"].copy()
reg = datasets["studentRegistration"].copy()

# Convert '?' to missing values
reg["date_unregistration_clean"] = (
    reg["date_unregistration"].replace("?", pd.NA)
)

reg["date_unregistration_clean"] = pd.to_numeric(
    reg["date_unregistration_clean"],
    errors="coerce"
)

reg["date_registration"] = pd.to_numeric(
    reg["date_registration"],
    errors="coerce"
)

# Merge student information and registration
student_status = info.merge(
    reg[[
        "code_module",
        "code_presentation",
        "id_student",
        "date_registration",
        "date_unregistration_clean"
    ]],
    on=["code_module", "code_presentation", "id_student"],
    how="left",
    validate="one_to_one"
)

# Investigate inconsistent cases
print("===== CASE 1: WITHDRAWN WITHOUT DATE =====")

unknown_withdrawal = student_status[
    (student_status["final_result"] == "Withdrawn") &
    (student_status["date_unregistration_clean"].isna())
]

print("Count:", len(unknown_withdrawal))
display(unknown_withdrawal.head(10))

print("\n===== CASE 2: NON-WITHDRAWN WITH UNREGISTRATION DATE =====")

inconsistent_records = student_status[
    (student_status["final_result"] != "Withdrawn") &
    (student_status["date_unregistration_clean"].notna())
]

print("Count:", len(inconsistent_records))
display(inconsistent_records)

# Eligibility conditions
registered_by_day28 = (
    student_status["date_registration"] <= PREDICTION_DAY
)

not_withdrawn_by_day28 = (
    student_status["date_unregistration_clean"].isna() |
    (student_status["date_unregistration_clean"] > PREDICTION_DAY)
)

known_withdrawal_timing = ~(
    (student_status["final_result"] == "Withdrawn") &
    (student_status["date_unregistration_clean"].isna())
)

# Exclusion reasons
mask_unknown = ~known_withdrawal_timing

mask_late_registration = (
    known_withdrawal_timing &
    ~registered_by_day28
)

mask_early_withdrawal = (
    known_withdrawal_timing &
    registered_by_day28 &
    ~not_withdrawn_by_day28
)

mask_eligible = (
    known_withdrawal_timing &
    registered_by_day28 &
    not_withdrawn_by_day28
)

print("\n===== DAY-28 COHORT SUMMARY =====")
print("Original records:", len(student_status))
print("Unknown withdrawal timing:", mask_unknown.sum())
print("Registered after day 28:", mask_late_registration.sum())
print("Unregistered on/before day 28:", mask_early_withdrawal.sum())
print("Eligible at day 28:", mask_eligible.sum())

# Create eligible cohort
cohort = student_status.loc[mask_eligible].copy()

# Binary target: 1 = eventual withdrawal, 0 = other final result
cohort["target"] = (
    cohort["final_result"] == "Withdrawn"
).astype(int)

print("\n===== TARGET DISTRIBUTION IN COHORT =====")
print(cohort["target"].value_counts())
print("\nTarget percentages:")
print(
    (cohort["target"].value_counts(normalize=True) * 100)
    .round(2)
)

print("\nFinal cohort shape:", cohort.shape)
```

```python

# ============================================================
# STEP 5: TEMPORAL FEATURE ENGINEERING (DAY 0-28)
# ============================================================

import pandas as pd
import numpy as np

# 1. Load VLE interactions
vle = datasets["studentVle"].copy()

# 2. Clean columns
vle["date"] = pd.to_numeric(vle["date"], errors="coerce")
vle["sum_click"] = pd.to_numeric(
    vle["sum_click"], errors="coerce"
)

vle = vle.dropna(
    subset=["date", "sum_click", "id_student"]
)

# 3. Keep only the observation window
vle = vle[
    (vle["date"] >= 0) &
    (vle["date"] <= 28)
].copy()

# 4. Keep interactions belonging to eligible students
keys = [
    "code_module",
    "code_presentation",
    "id_student"
]

cohort_keys = cohort[keys].drop_duplicates()

vle = vle.merge(
    cohort_keys,
    on=keys,
    how="inner",
    validate="many_to_one"
)

# 5. Remove exact duplicate records only
before = len(vle)
vle = vle.drop_duplicates()
after = len(vle)

print("Exact duplicate rows removed:", before - after)
print("VLE rows after filtering:", after)

# 6. Define weekly windows
vle["week"] = pd.cut(
    vle["date"],
    bins=[-1, 6, 13, 20, 28],
    labels=[1, 2, 3, 4]
).astype(int)

# 7. Aggregate weekly features
weekly = (
    vle.groupby(keys + ["week"], observed=True)
    .agg(
        weekly_clicks=("sum_click", "sum"),
        weekly_active_days=("date", "nunique"),
        weekly_active_sites=("id_site", "nunique")
    )
    .reset_index()
)

# 8. Convert weekly features to wide format
weekly_features = weekly.pivot(
    index=keys,
    columns="week",
    values=[
        "weekly_clicks",
        "weekly_active_days",
        "weekly_active_sites"
    ]
)

weekly_features.columns = [
    f"{feature}_week{week}"
    for feature, week in weekly_features.columns
]

weekly_features = weekly_features.reset_index()

# 9. Overall temporal features
overall = (
    vle.groupby(keys)
    .agg(
        total_clicks=("sum_click", "sum"),
        active_days=("date", "nunique"),
        active_sites=("id_site", "nunique"),
        mean_daily_clicks=("sum_click", "mean"),
        first_activity_day=("date", "min"),
        last_activity_day=("date", "max")
    )
    .reset_index()
)

# 10. Merge all features with eligible cohort
temporal_features = cohort[
    keys + ["target", "final_result"]
].merge(
    overall,
    on=keys,
    how="left",
    validate="one_to_one"
).merge(
    weekly_features,
    on=keys,
    how="left",
    validate="one_to_one"
)

# 11. Students without VLE activity
feature_columns = [
    c for c in temporal_features.columns
    if c not in keys + ["target", "final_result"]
]

temporal_features[feature_columns] = (
    temporal_features[feature_columns].fillna(0)
)

# 12. Summary
print("\n========== TEMPORAL FEATURE SUMMARY ==========")

print("Number of students:", len(temporal_features))
print("Number of features:", len(feature_columns))

print("\nDataset shape:", temporal_features.shape)

print("\nTarget distribution:")
print(temporal_features["target"].value_counts())

print("\nMissing values:")
print(temporal_features.isna().sum().sum())

print("\nFirst rows:")
display(temporal_features.head())

# 13. Save dataset
temporal_features.to_csv(
    "/content/day28_temporal_features.csv",
    index=False
)

print("\nSaved: day28_temporal_features.csv")
```

```python

# ============================================================
# STEP 5.1: DATA QUALITY AUDIT + BEHAVIORAL FEATURES
# ============================================================

import pandas as pd
import numpy as np

keys = ["code_module", "code_presentation", "id_student"]

# ------------------------------------------------------------
# A. DUPLICATE SENSITIVITY AUDIT
# ------------------------------------------------------------

raw = datasets["studentVle"].copy()

raw["date"] = pd.to_numeric(raw["date"], errors="coerce")
raw["sum_click"] = pd.to_numeric(raw["sum_click"], errors="coerce")

raw = raw.dropna(subset=["date", "sum_click"])

raw = raw[
    raw["date"].between(0, 28)
].merge(
    cohort[keys].drop_duplicates(),
    on=keys,
    how="inner"
)

duplicate_count = raw.duplicated().sum()

raw_clicks = raw.groupby(keys)["sum_click"].sum()

clean = raw.drop_duplicates()

clean_clicks = clean.groupby(keys)["sum_click"].sum()

comparison = pd.DataFrame({
    "raw_clicks": raw_clicks,
    "clean_clicks": clean_clicks
}).fillna(0)

comparison["difference"] = (
    comparison["raw_clicks"] - comparison["clean_clicks"]
)

print("\n========== DUPLICATE SENSITIVITY ==========")
print("Rows before deduplication:", len(raw))
print("Exact duplicate rows:", duplicate_count)
print("Rows after deduplication:", len(clean))
print("Total clicks before:", raw["sum_click"].sum())
print("Total clicks after:", clean["sum_click"].sum())
print("Students affected:", (comparison["difference"] > 0).sum())
print("Clicks removed:", comparison["difference"].sum())

# ------------------------------------------------------------
# B. CORRECT DAILY ACTIVITY FEATURES
# ------------------------------------------------------------

daily = (
    clean.groupby(keys + ["date"])
    .agg(
        daily_clicks=("sum_click", "sum"),
        daily_sites=("id_site", "nunique")
    )
    .reset_index()
)

daily_features = (
    daily.groupby(keys)
    .agg(
        mean_clicks_per_active_day=("daily_clicks", "mean"),
        max_daily_clicks=("daily_clicks", "max"),
        daily_click_std=("daily_clicks", "std")
    )
    .reset_index()
)

# ------------------------------------------------------------
# C. BEHAVIORAL AND TEMPORAL FEATURES
# ------------------------------------------------------------

behavior = weekly_features.copy()

click_cols = [
    f"weekly_clicks_week{i}" for i in range(1, 5)
]

# Replace missing weekly values
behavior[click_cols] = behavior[click_cols].fillna(0)

# Weekly activity trend
behavior["weekly_clicks_slope"] = np.polyfit(
    [1, 2, 3, 4],
    behavior[click_cols].to_numpy().T,
    1
)[0]

behavior["week4_minus_week1"] = (
    behavior["weekly_clicks_week4"]
    - behavior["weekly_clicks_week1"]
)

behavior["activity_decline"] = (
    behavior["weekly_clicks_week1"]
    > behavior["weekly_clicks_week4"]
).astype(int)

# Merge features
enhanced_features = temporal_features.merge(
    daily_features,
    on=keys,
    how="left",
    validate="one_to_one"
).merge(
    behavior[
        keys + [
            "weekly_clicks_slope",
            "week4_minus_week1",
            "activity_decline"
        ]
    ],
    on=keys,
    how="left",
    validate="one_to_one"
)

# No activity indicator
enhanced_features["no_vle_activity"] = (
    enhanced_features["total_clicks"] == 0
).astype(int)

# Inactivity gap at prediction time
enhanced_features["days_since_last_activity"] = (
    28 - enhanced_features["last_activity_day"]
)

# Fill features for students without activity
new_features = [
    "mean_clicks_per_active_day",
    "max_daily_clicks",
    "daily_click_std",
    "weekly_clicks_slope",
    "week4_minus_week1",
    "activity_decline",
    "no_vle_activity",
    "days_since_last_activity"
]

enhanced_features[new_features] = (
    enhanced_features[new_features].fillna(0)
)

print("\n========== ENHANCED FEATURE SUMMARY ==========")
print("Shape:", enhanced_features.shape)
print("New features:", new_features)

print("\nActivity decline distribution:")
print(enhanced_features["activity_decline"].value_counts())

print("\nNo activity:")
print(enhanced_features["no_vle_activity"].value_counts())

print("\nMissing values:", enhanced_features.isna().sum().sum())

enhanced_features.to_csv(
    "/content/day28_enhanced_features.csv",
    index=False
)

print("\nSaved: day28_enhanced_features.csv")
```

```python

# ============================================================
# STEP 5.2: CANONICAL STUDENT-VLE INTERACTIONS
# ============================================================

# Start from the exact-deduplicated data
vle_canonical = clean.copy()

interaction_key = [
    "code_module",
    "code_presentation",
    "id_student",
    "id_site",
    "date"
]

# Aggregate repeated student-resource-day records
vle_canonical = (
    vle_canonical
    .groupby(interaction_key, as_index=False)
    .agg(sum_click=("sum_click", "sum"))
)

# Validate unique interaction keys
print("========== CANONICAL VLE VALIDATION ==========")

print("Rows:", len(vle_canonical))

print(
    "Unique interaction keys:",
    vle_canonical[interaction_key].drop_duplicates().shape[0]
)

print(
    "Duplicate interaction keys:",
    vle_canonical.duplicated(interaction_key).sum()
)

print(
    "Total clicks:",
    vle_canonical["sum_click"].sum()
)

print(
    "Number of students:",
    vle_canonical["id_student"].nunique()
)

assert not vle_canonical.duplicated(interaction_key).any()

# Save canonical table
vle_canonical.to_csv(
    "/content/studentVLE_day28_canonical.csv",
    index=False
)

print("\nCanonical VLE dataset saved.")
```

```python

# ============================================================
# STEP 6: VALIDATE STUDENT-COURSE ACTIVITY
# ============================================================

keys = [
    "code_module",
    "code_presentation",
    "id_student"
]

# Unique student-course registrations with VLE activity
active_enrollments = (
    vle_canonical[keys]
    .drop_duplicates()
)

# All eligible enrollments
all_enrollments = cohort[keys].drop_duplicates()

# Identify students without VLE activity
activity_check = all_enrollments.merge(
    active_enrollments.assign(has_activity=1),
    on=keys,
    how="left",
    validate="one_to_one"
)

activity_check["has_activity"] = (
    activity_check["has_activity"].fillna(0).astype(int)
)

print("\n========== STUDENT ACTIVITY VALIDATION ==========")

print("Eligible enrollments:", len(all_enrollments))

print(
    "Active enrollments:",
    activity_check["has_activity"].sum()
)

print(
    "No-activity enrollments:",
    (activity_check["has_activity"] == 0).sum()
)

print(
    "Activity rate (%):",
    round(activity_check["has_activity"].mean() * 100, 2)
)

# Compare with previously generated features
comparison = temporal_features[keys + ["no_vle_activity"]].merge(
    activity_check,
    on=keys,
    how="left",
    validate="one_to_one"
)

print("\nAgreement between activity indicators:")
print(
    pd.crosstab(
        comparison["no_vle_activity"],
        comparison["has_activity"],
        rownames=["Previous indicator"],
        colnames=["Canonical indicator"]
    )
)

# Verify enrollment-level uniqueness
assert not activity_check.duplicated(keys).any()

print("\nValidation completed.")
```

```python

# ============================================================
# STEP 7: STUDENT-RESOURCE BIPARTITE GRAPH
# ============================================================

import pandas as pd
import numpy as np

student_keys = [
    "code_module",
    "code_presentation",
    "id_student"
]

resource_keys = [
    "code_module",
    "code_presentation",
    "id_site"
]

# ------------------------------------------------------------
# 1. Student nodes: preserve all eligible enrollments
# ------------------------------------------------------------

student_nodes = (
    cohort[student_keys]
    .drop_duplicates()
    .sort_values(student_keys)
    .reset_index(drop=True)
)

student_nodes["student_node_id"] = np.arange(
    len(student_nodes)
)

student_nodes["node_type"] = "student"

# ------------------------------------------------------------
# 2. Resource nodes: course-specific resources
# ------------------------------------------------------------

resource_nodes = (
    vle_canonical[resource_keys]
    .drop_duplicates()
    .sort_values(resource_keys)
    .reset_index(drop=True)
)

resource_nodes["resource_node_id"] = (
    np.arange(len(resource_nodes))
    + len(student_nodes)
)

resource_nodes["node_type"] = "resource"

# ------------------------------------------------------------
# 3. Define temporal windows
# ------------------------------------------------------------

vle_graph = vle_canonical.copy()

vle_graph["week"] = pd.cut(
    vle_graph["date"],
    bins=[-1, 6, 13, 20, 28],
    labels=[1, 2, 3, 4]
).astype(int)

# ------------------------------------------------------------
# 4. Aggregate student-resource interactions
# ------------------------------------------------------------

edge_base = (
    vle_graph.groupby(
        student_keys + resource_keys[2:],
        as_index=False
    )
    .agg(
        total_clicks=("sum_click", "sum"),
        active_days=("date", "nunique"),
        first_day=("date", "min"),
        last_day=("date", "max")
    )
)

# Weekly interaction weights
weekly_edges = (
    vle_graph.groupby(
        student_keys + resource_keys[2:] + ["week"],
        observed=True
    )["sum_click"]
    .sum()
    .unstack("week", fill_value=0)
    .reset_index()
)

weekly_edges.columns = [
    f"week{int(c)}_clicks" if isinstance(c, (int, np.integer))
    else c
    for c in weekly_edges.columns
]

for w in range(1, 5):
    col = f"week{w}_clicks"
    if col not in weekly_edges.columns:
        weekly_edges[col] = 0

# ------------------------------------------------------------
# 5. Add node IDs to edges
# ------------------------------------------------------------

edges = edge_base.merge(
    student_nodes[
        student_keys + ["student_node_id"]
    ],
    on=student_keys,
    how="left",
    validate="many_to_one"
)

edges = edges.merge(
    resource_nodes[
        resource_keys + ["resource_node_id"]
    ],
    on=resource_keys,
    how="left",
    validate="many_to_one"
)

edges = edges.merge(
    weekly_edges,
    on=student_keys + resource_keys[2:],
    how="left",
    validate="one_to_one"
)

# ------------------------------------------------------------
# 6. Create bidirectional graph edges
# ------------------------------------------------------------

forward = edges[
    [
        "student_node_id",
        "resource_node_id",
        "total_clicks",
        "active_days",
        "first_day",
        "last_day",
        "week1_clicks",
        "week2_clicks",
        "week3_clicks",
        "week4_clicks"
    ]
].copy()

forward["source"] = forward["student_node_id"]
forward["destination"] = forward["resource_node_id"]
forward["relation"] = "student_to_resource"

reverse = forward.copy()

reverse["source"] = forward["destination"]
reverse["destination"] = forward["source"]
reverse["relation"] = "resource_to_student"

graph_edges = pd.concat(
    [forward, reverse],
    ignore_index=True
)

# ------------------------------------------------------------
# 7. Graph validation
# ------------------------------------------------------------

print("\n========== BIPARTITE GRAPH SUMMARY ==========")

print("Student nodes:", len(student_nodes))
print("Resource nodes:", len(resource_nodes))
print("Unique interaction pairs:", len(edges))
print("Directed graph edges:", len(graph_edges))

print(
    "Total interaction clicks:",
    edges["total_clicks"].sum()
)

print(
    "Students without edges:",
    len(student_nodes) -
    edges["student_node_id"].nunique()
)

print(
    "Missing student IDs:",
    edges["student_node_id"].isna().sum()
)

print(
    "Missing resource IDs:",
    edges["resource_node_id"].isna().sum()
)

# Assertions
assert edges["student_node_id"].notna().all()
assert edges["resource_node_id"].notna().all()

assert len(edges) == edges[
    ["student_node_id", "resource_node_id"]
].drop_duplicates().shape[0]

assert edges["total_clicks"].sum() == (
    vle_canonical["sum_click"].sum()
)

print("\nGraph validation completed successfully.")

# ------------------------------------------------------------
# 8. Save graph files
# ------------------------------------------------------------

student_nodes.to_csv(
    "/content/student_nodes.csv",
    index=False
)

resource_nodes.to_csv(
    "/content/resource_nodes.csv",
    index=False
)

edges.to_csv(
    "/content/student_resource_edges.csv",
    index=False
)

graph_edges.to_csv(
    "/content/bipartite_graph_edges.csv",
    index=False
)

print("\nAll graph files saved successfully.")
```

```python

from sklearn.model_selection import StratifiedGroupKFold
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    precision_score, recall_score, f1_score,
    balanced_accuracy_score, roc_auc_score,
    average_precision_score
)
import numpy as np
import pandas as pd

X = proposed_df[proposed_cols].apply(
    pd.to_numeric, errors="coerce"
).reset_index(drop=True)

y = proposed_df["target"].astype(int).reset_index(drop=True)
groups = proposed_df["id_student"].reset_index(drop=True)

thresholds = np.arange(0.05, 0.51, 0.01)

def build_model():
    return Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("classifier", RandomForestClassifier(
            n_estimators=300,
            random_state=42,
            class_weight="balanced",
            n_jobs=-1
        ))
    ])

outer_cv = StratifiedGroupKFold(
    n_splits=5, shuffle=True, random_state=42
)

outer_predictions = np.zeros(len(y), dtype=int)
outer_probabilities = np.zeros(len(y))
selected_thresholds = []

for fold, (train_idx, test_idx) in enumerate(
    outer_cv.split(X, y, groups), start=1
):

    X_train = X.iloc[train_idx]
    y_train = y.iloc[train_idx]
    g_train = groups.iloc[train_idx]

    # Inner CV: select threshold using training data only
    inner_cv = StratifiedGroupKFold(
        n_splits=4, shuffle=True, random_state=42
    )

    inner_prob = np.zeros(len(train_idx))

    for inner_train, inner_valid in inner_cv.split(
        X_train, y_train, g_train
    ):
        model = build_model()

        model.fit(
            X_train.iloc[inner_train],
            y_train.iloc[inner_train]
        )

        inner_prob[inner_valid] = model.predict_proba(
            X_train.iloc[inner_valid]
        )[:, 1]

    f1_values = [
        f1_score(
            y_train,
            (inner_prob >= t).astype(int),
            zero_division=0
        )
        for t in thresholds
    ]

    best_threshold = float(thresholds[np.argmax(f1_values)])
    selected_thresholds.append(best_threshold)

    # Outer test: independent evaluation
    final_model = build_model()
    final_model.fit(X_train, y_train)

    test_prob = final_model.predict_proba(
        X.iloc[test_idx]
    )[:, 1]

    outer_probabilities[test_idx] = test_prob
    outer_predictions[test_idx] = (
        test_prob >= best_threshold
    ).astype(int)

    print(
        f"Fold {fold}: selected threshold = "
        f"{best_threshold:.2f}"
    )

# Final independent outer-CV evaluation
print("\n===== NESTED GROUP CV RESULTS =====")

print("Precision:",
      precision_score(y, outer_predictions, zero_division=0))

print("Recall:",
      recall_score(y, outer_predictions, zero_division=0))

print("F1-score:",
      f1_score(y, outer_predictions, zero_division=0))

print("Balanced Accuracy:",
      balanced_accuracy_score(y, outer_predictions))

print("ROC-AUC:",
      roc_auc_score(y, outer_probabilities))

print("PR-AUC:",
      average_precision_score(y, outer_probabilities))

print("Selected thresholds:", selected_thresholds)
print("Mean threshold:", np.mean(selected_thresholds))

nested_results = pd.DataFrame({
    "Actual": y,
    "Predicted": outer_predictions,
    "Probability": outer_probabilities
})

nested_results.to_csv(
    "/content/nested_cv_predictions.csv",
    index=False
)
```

```python

# ============================================================
# EXPERIMENT 2: NESTED GROUP CV ABLATION STUDY
# A1: Baseline
# A2: Proposed
# A3: Proposed + Isolation Forest
# A4: Proposed + SMOTE
# A5: Proposed + Isolation Forest + SMOTE
# ============================================================

import time
import numpy as np
import pandas as pd

from sklearn.model_selection import StratifiedGroupKFold
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier, IsolationForest
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    balanced_accuracy_score
)

from imblearn.over_sampling import SMOTE

# ------------------------------------------------------------
# 1. Prepare features and target
# ------------------------------------------------------------

keys = ["code_module", "code_presentation", "id_student"]

excluded = keys + ["target", "final_result"]

baseline_cols = [
    c for c in temporal_features.columns
    if c not in excluded
]

proposed_cols = [
    c for c in enhanced_features.columns
    if c not in excluded
]

X_baseline = temporal_features[baseline_cols].apply(
    pd.to_numeric, errors="coerce"
).reset_index(drop=True)

X_proposed = enhanced_features[proposed_cols].apply(
    pd.to_numeric, errors="coerce"
).reset_index(drop=True)

y = enhanced_features["target"].astype(int).reset_index(drop=True)

groups = enhanced_features["id_student"].reset_index(drop=True)

assert len(X_baseline) == len(X_proposed) == len(y)
assert y.notna().all()

print("Dataset size:", len(y))
print("Baseline features:", len(baseline_cols))
print("Proposed features:", len(proposed_cols))
print("Target distribution:")
print(y.value_counts())

# ------------------------------------------------------------
# 2. Define configurations
# ------------------------------------------------------------

configurations = {
    "A1_Baseline": {
        "X": X_baseline,
        "use_if": False,
        "use_smote": False
    },
    "A2_Proposed": {
        "X": X_proposed,
        "use_if": False,
        "use_smote": False
    },
    "A3_Proposed_IF": {
        "X": X_proposed,
        "use_if": True,
        "use_smote": False
    },
    "A4_Proposed_SMOTE": {
        "X": X_proposed,
        "use_if": False,
        "use_smote": True
    },
    "A5_Proposed_IF_SMOTE": {
        "X": X_proposed,
        "use_if": True,
        "use_smote": True
    }
}

threshold_grid = np.arange(0.05, 0.51, 0.01)

# Same model settings for all configurations
N_ESTIMATORS = 200
RANDOM_STATE = 42

# ------------------------------------------------------------
# 3. Training helper
# ------------------------------------------------------------

def fit_and_predict(
    X_train,
    y_train,
    X_valid,
    use_if=False,
    use_smote=False
):

    # Imputation fitted only on training data
    imputer = SimpleImputer(strategy="median")

    Xtr = imputer.fit_transform(X_train)
    Xva = imputer.transform(X_valid)

    ytr = np.asarray(y_train)

    removed_outliers = 0

    # Isolation Forest: training data only
    if use_if:

        detector = IsolationForest(
            n_estimators=100,
            contamination=0.02,
            random_state=RANDOM_STATE,
            n_jobs=-1
        )

        inlier_mask = detector.fit_predict(Xtr) == 1

        removed_outliers = int((~inlier_mask).sum())

        Xtr = Xtr[inlier_mask]
        ytr = ytr[inlier_mask]

    # SMOTE: training data only
    if use_smote:

        class_counts = pd.Series(ytr).value_counts()

        if len(class_counts) < 2:
            raise ValueError(
                "Only one class remains after training preprocessing."
            )

        minority_count = int(class_counts.min())

        if minority_count < 2:
            raise ValueError(
                "Not enough minority samples for SMOTE."
            )

        k = min(5, minority_count - 1)

        sampler = SMOTE(
            random_state=RANDOM_STATE,
            k_neighbors=k
        )

        Xtr, ytr = sampler.fit_resample(Xtr, ytr)

    # Same classifier settings across all configurations
    model = RandomForestClassifier(
        n_estimators=N_ESTIMATORS,
        random_state=RANDOM_STATE,
        class_weight=None,
        n_jobs=-1
    )

    model.fit(Xtr, ytr)

    probabilities = model.predict_proba(Xva)[:, 1]

    return probabilities, removed_outliers


# ------------------------------------------------------------
# 4. Create identical outer folds
# ------------------------------------------------------------

outer_cv = StratifiedGroupKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

outer_splits = list(
    outer_cv.split(X_proposed, y, groups)
)

# ------------------------------------------------------------
# 5. Run nested CV for each configuration
# ------------------------------------------------------------

all_results = []

for config_name, config in configurations.items():

    print("\n" + "=" * 65)
    print("Running:", config_name)
    print("=" * 65)

    X = config["X"]

    use_if = config["use_if"]
    use_smote = config["use_smote"]

    oof_prob = np.zeros(len(y))
    oof_pred = np.zeros(len(y), dtype=int)

    selected_thresholds = []
    removed_counts = []

    start_time = time.time()

    for fold, (outer_train, outer_test) in enumerate(
        outer_splits, start=1
    ):

        X_outer_train = X.iloc[outer_train]
        y_outer_train = y.iloc[outer_train]
        g_outer_train = groups.iloc[outer_train]

        # Inner CV for threshold selection
        inner_cv = StratifiedGroupKFold(
            n_splits=4,
            shuffle=True,
            random_state=42
        )

        inner_splits = list(
            inner_cv.split(
                X_outer_train,
                y_outer_train,
                g_outer_train
            )
        )

        inner_prob = np.zeros(len(outer_train))

        for inner_train, inner_valid in inner_splits:

            probabilities, _ = fit_and_predict(
                X_outer_train.iloc[inner_train],
                y_outer_train.iloc[inner_train],
                X_outer_train.iloc[inner_valid],
                use_if=use_if,
                use_smote=use_smote
            )

            inner_prob[inner_valid] = probabilities

        # Select threshold using inner OOF predictions
        f1_values = []

        for threshold in threshold_grid:

            pred = (inner_prob >= threshold).astype(int)

            f1_values.append(
                f1_score(
                    y_outer_train,
                    pred,
                    zero_division=0
                )
            )

        best_threshold = float(
            threshold_grid[np.argmax(f1_values)]
        )

        selected_thresholds.append(best_threshold)

        # Train on complete outer training set
        # Evaluate on untouched outer test set
        test_prob, removed = fit_and_predict(
            X.iloc[outer_train],
            y.iloc[outer_train],
            X.iloc[outer_test],
            use_if=use_if,
            use_smote=use_smote
        )

        removed_counts.append(removed)

        oof_prob[outer_test] = test_prob

        oof_pred[outer_test] = (
            test_prob >= best_threshold
        ).astype(int)

        print(
            f"Fold {fold}: threshold={best_threshold:.2f}, "
            f"removed outliers={removed}"
        )

    elapsed = time.time() - start_time

    # Overall outer-fold performance
    result = {
        "Configuration": config_name,
        "Features": X.shape[1],
        "Precision": precision_score(
            y, oof_pred, zero_division=0
        ),
        "Recall": recall_score(
            y, oof_pred, zero_division=0
        ),
        "F1-score": f1_score(
            y, oof_pred, zero_division=0
        ),
        "Balanced_Accuracy": balanced_accuracy_score(
            y, oof_pred
        ),
        "ROC-AUC": roc_auc_score(y, oof_prob),
        "PR-AUC": average_precision_score(y, oof_prob),
        "Mean_Threshold": np.mean(selected_thresholds),
        "Execution_Time_sec": elapsed,
        "Mean_IF_Removed": np.mean(removed_counts)
    }

    all_results.append(result)

    print("\nConfiguration results:")
    print(pd.Series(result).to_string())


# ------------------------------------------------------------
# 6. Display and save final ablation table
# ------------------------------------------------------------

ablation_results = pd.DataFrame(all_results)

print("\n\n========== FINAL ABLATION STUDY ==========")

display(
    ablation_results.round(4)
)

ablation_results.to_csv(
    "/content/ablation_study_results.csv",
    index=False
)

print("\nSaved: /content/ablation_study_results.csv")
```

```python

# ============================================================
# DATA QUALITY ASSESSMENT - OULAD
# Diagnostic Stage
# ============================================================

import pandas as pd
import numpy as np

def quality_report(data, name):

    data = pd.DataFrame(data).copy()

    total_cells = data.size

    missing = data.isna().sum().sum()

    duplicates = data.duplicated().sum()

    numeric = data.select_dtypes(include=np.number)

    outlier_rows = set()

    for col in numeric.columns:

        s = numeric[col].dropna()

        if len(s) < 4:
            continue

        q1 = s.quantile(0.25)
        q3 = s.quantile(0.75)
        iqr = q3 - q1

        if iqr == 0:
            continue

        lower = q1 - 1.5 * iqr
        upper = q3 + 1.5 * iqr

        indexes = data.index[
            (data[col] < lower) |
            (data[col] > upper)
        ]

        outlier_rows.update(indexes)

    return {
        "Dataset": name,
        "Rows": len(data),
        "Columns": data.shape[1],
        "Missing_Count": int(missing),
        "Missing_Percentage": round(
            missing / total_cells * 100, 4
        ) if total_cells else 0,
        "Duplicate_Rows": int(duplicates),
        "Duplicate_Percentage": round(
            duplicates / len(data) * 100, 4
        ) if len(data) else 0,
        "Outlier_Rows": len(outlier_rows),
        "Outlier_Percentage": round(
            len(outlier_rows) / len(data) * 100, 4
        ) if len(data) else 0
    }


# Assess available datasets
reports = []

for name in ["df", "baseline_df", "proposed_df",
             "X_baseline", "X_proposed",
             "enhanced_features"]:

    if name in globals():

        reports.append(
            quality_report(globals()[name], name)
        )

quality_table = pd.DataFrame(reports)

print("========== DATA QUALITY DIAGNOSTICS ==========")

display(quality_table)

quality_table.to_csv(
    "/content/data_quality_diagnostics.csv",
    index=False
)

print("Saved: /content/data_quality_diagnostics.csv")
```

```python

# ============================================================
# SCHEMA VALIDATION AND DATA CONSISTENCY
# JCOMSS - OULAD Dataset
# ============================================================

import pandas as pd
import numpy as np

data = df.copy()

results = []

def add_check(name, violations, details=""):
    results.append({
        "Validation_Check": name,
        "Violations": int(violations),
        "Status": "PASS" if violations == 0 else "CHECK",
        "Details": details
    })

# ------------------------------------------------------------
# 1. Required columns
# ------------------------------------------------------------

required_columns = [
    "code_module", "code_presentation", "id_student",
    "target", "final_result", "total_clicks",
    "active_days", "active_sites",
    "first_activity_day", "last_activity_day",
    "weekly_clicks_week1", "weekly_clicks_week2",
    "weekly_clicks_week3", "weekly_clicks_week4"
]

missing_columns = [
    c for c in required_columns if c not in data.columns
]

add_check(
    "Required Columns",
    len(missing_columns),
    str(missing_columns)
)

# ------------------------------------------------------------
# 2. Unique learner-module-presentation key
# ------------------------------------------------------------

key_columns = [
    "code_module", "code_presentation", "id_student"
]

duplicate_keys = data.duplicated(
    subset=key_columns,
    keep=False
).sum()

add_check(
    "Duplicate Enrollment Keys",
    duplicate_keys,
    "Key: module + presentation + student"
)

# ------------------------------------------------------------
# 3. Target validation
# ------------------------------------------------------------

invalid_targets = (
    ~data["target"].isin([0, 1])
).sum()

add_check(
    "Binary Target Validity",
    invalid_targets
)

# Check consistency between target and final result
target_mismatch = (
    data["target"] !=
    (data["final_result"] == "Withdrawn").astype(int)
).sum()

add_check(
    "Target-Final Result Consistency",
    target_mismatch
)

# ------------------------------------------------------------
# 4. Categorical values
# ------------------------------------------------------------

allowed_results = [
    "Pass", "Distinction", "Fail", "Withdrawn"
]

invalid_results = (
    ~data["final_result"].isin(allowed_results)
).sum()

add_check(
    "Final Result Categories",
    invalid_results,
    "Expected: Pass, Distinction, Fail, Withdrawn"
)

# ------------------------------------------------------------
# 5. Non-negative activity features
# ------------------------------------------------------------

nonnegative_columns = [
    "total_clicks",
    "active_days",
    "active_sites",
    "weekly_clicks_week1",
    "weekly_clicks_week2",
    "weekly_clicks_week3",
    "weekly_clicks_week4",
    "weekly_active_days_week1",
    "weekly_active_days_week2",
    "weekly_active_days_week3",
    "weekly_active_days_week4"
]

for col in nonnegative_columns:

    if col in data.columns:

        violations = (
            data[col].dropna() < 0
        ).sum()

        add_check(
            f"Non-negative: {col}",
            violations
        )

# ------------------------------------------------------------
# 6. Weekly active days: maximum 7 days
# ------------------------------------------------------------

weekly_day_columns = [
    "weekly_active_days_week1",
    "weekly_active_days_week2",
    "weekly_active_days_week3",
    "weekly_active_days_week4"
]

for col in weekly_day_columns:

    if col in data.columns:

        violations = (
            (data[col] < 0) |
            (data[col] > 7)
        ).sum()

        add_check(
            f"Weekly Days Range: {col}",
            violations
        )

# ------------------------------------------------------------
# 7. Temporal consistency
# ------------------------------------------------------------

if all(c in data.columns for c in [
    "first_activity_day", "last_activity_day"
]):

    invalid_temporal = (
        data["first_activity_day"] >
        data["last_activity_day"]
    ).sum()

    add_check(
        "First/Last Activity Consistency",
        invalid_temporal
    )

# ------------------------------------------------------------
# 8. Activity consistency
# ------------------------------------------------------------

if "no_vle_activity" in data.columns:

    no_activity_mismatch = (
        (data["no_vle_activity"] == 1) &
        (data["total_clicks"] > 0)
    ).sum()

    add_check(
        "No-VLE Activity Consistency",
        no_activity_mismatch
    )

# ------------------------------------------------------------
# 9. Final report
# ------------------------------------------------------------

schema_report = pd.DataFrame(results)

print("========== SCHEMA VALIDATION ==========")
display(schema_report)

print("\nTotal validation checks:", len(schema_report))
print("Passed checks:",
      (schema_report["Status"] == "PASS").sum())
print("Checks requiring inspection:",
      (schema_report["Status"] == "CHECK").sum())

schema_report.to_csv(
    "/content/schema_validation_results.csv",
    index=False
)

print("\nSaved: /content/schema_validation_results.csv")
```

```python

# ============================================================
# A0: NESTED GROUP CROSS-VALIDATION
# RAW MERGE - NO DEDUPLICATION - NO IF - NO SMOTE
# ============================================================

import time
import numpy as np

from sklearn.model_selection import StratifiedGroupKFold
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    roc_auc_score,
    average_precision_score
)

print("=" * 70)
print("RUNNING A0 - RAW MERGE, NO DEDUPLICATION")
print("=" * 70)

start_time = time.time()

# Identical evaluation protocol
OUTER_SPLITS = 5
INNER_SPLITS = 4

threshold_grid = np.arange(0.05, 0.51, 0.01)

outer_cv = StratifiedGroupKFold(
    n_splits=OUTER_SPLITS,
    shuffle=True,
    random_state=42
)

results = []
fold_thresholds = []

X = X_A0.reset_index(drop=True)
y = y_A0.reset_index(drop=True)
groups = groups_A0.reset_index(drop=True)

# Random Forest configuration
def create_model():
    return Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("rf", RandomForestClassifier(
            n_estimators=200,
            random_state=42,
            class_weight=None,
            n_jobs=-1
        ))
    ])

# Outer evaluation
for fold, (train_idx, test_idx) in enumerate(
    outer_cv.split(X, y, groups), start=1
):

    print(f"\nOuter fold {fold}/{OUTER_SPLITS}")

    X_train = X.iloc[train_idx]
    X_test = X.iloc[test_idx]

    y_train = y.iloc[train_idx]
    y_test = y.iloc[test_idx]

    g_train = groups.iloc[train_idx]

    # -----------------------------------------
    # INNER CV: threshold selection
    # -----------------------------------------

    inner_cv = StratifiedGroupKFold(
        n_splits=INNER_SPLITS,
        shuffle=True,
        random_state=42
    )

    inner_oof = np.zeros(len(X_train))

    for inner_train_idx, inner_val_idx in inner_cv.split(
        X_train, y_train, g_train
    ):

        model = create_model()

        model.fit(
            X_train.iloc[inner_train_idx],
            y_train.iloc[inner_train_idx]
        )

        inner_oof[inner_val_idx] = model.predict_proba(
            X_train.iloc[inner_val_idx]
        )[:, 1]

    # Select threshold using inner OOF predictions
    threshold_scores = []

    for threshold in threshold_grid:

        pred = (inner_oof >= threshold).astype(int)

        threshold_scores.append(
            f1_score(y_train, pred, zero_division=0)
        )

    best_threshold = threshold_grid[
        np.argmax(threshold_scores)
    ]

    fold_thresholds.append(best_threshold)

    # -----------------------------------------
    # OUTER TRAINING
    # -----------------------------------------

    final_model = create_model()

    final_model.fit(X_train, y_train)

    probabilities = final_model.predict_proba(
        X_test
    )[:, 1]

    predictions = (
        probabilities >= best_threshold
    ).astype(int)

    # -----------------------------------------
    # OUTER TEST METRICS
    # -----------------------------------------

    fold_result = {
        "Fold": fold,

        "Precision": precision_score(
            y_test, predictions, zero_division=0
        ),

        "Recall": recall_score(
            y_test, predictions, zero_division=0
        ),

        "F1": f1_score(
            y_test, predictions, zero_division=0
        ),

        "Bal_acc": balanced_accuracy_score(
            y_test, predictions
        ),

        "ROC_AUC": roc_auc_score(
            y_test, probabilities
        ),

        "PR_AUC": average_precision_score(
            y_test, probabilities
        ),

        "Threshold": best_threshold
    }

    results.append(fold_result)

    print(
        f"Threshold={best_threshold:.2f} | "
        f"F1={fold_result['F1']:.4f} | "
        f"ROC-AUC={fold_result['ROC_AUC']:.4f}"
    )

# -----------------------------------------
# FINAL RESULTS
# -----------------------------------------

results_df = pd.DataFrame(results)

execution_time = time.time() - start_time

final_results = results_df.drop(
    columns=["Fold"]
).mean()

print("\n" + "=" * 70)
print("A0 FINAL RESULTS")
print("=" * 70)

print(results_df.to_string(index=False))

print("\nMean metrics:")
print(final_results)

print("\nMean threshold:", np.mean(fold_thresholds))
print("Fold thresholds:", fold_thresholds)

print("Execution time:", round(execution_time, 2), "seconds")
print("IF removed:", 0)

# Save results
results_df.to_csv(
    "/content/A0_fold_results.csv",
    index=False
)

final_results.to_csv(
    "/content/A0_final_results.csv"
)

print("\nA0 evaluation completed.")
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.impute import SimpleImputer
from sklearn.ensemble import IsolationForest

# ============================================================
# FIGURE 4: ISOLATION FOREST OUTLIER-SCREENING DIAGNOSTICS
# Day-28 OULAD cohort | contamination = 0.02
# ============================================================

print("=" * 70)
print("FIGURE 4 - ISOLATION FOREST DESCRIPTIVE DIAGNOSTICS")
print("=" * 70)

# 1. Prepare the feature matrix and final-result labels
X = X_proposed.copy()

labels = proposed_df["final_result"].copy()

# Ensure that rows are aligned
assert X.index.equals(proposed_df.index), "Index mismatch!"
assert labels.index.equals(X.index), "Label index mismatch!"

# Ensure numeric features
X = X.apply(pd.to_numeric, errors="coerce")

# 2. Handle missing values
imputer = SimpleImputer(strategy="median")
X_imputed = imputer.fit_transform(X)

# 3. Fit Isolation Forest on the full day-28 cohort
iso = IsolationForest(
    n_estimators=100,
    contamination=0.02,
    random_state=42,
    n_jobs=-1
)

predictions = iso.fit_predict(X_imputed)

# -1 = anomaly, 1 = inlier
anomaly = predictions == -1

# Higher score = more anomalous
anomaly_score = -iso.score_samples(X_imputed)

# 4. Build diagnostic dataframe
diagnostics = pd.DataFrame({
    "final_result": labels.to_numpy(),
    "is_anomaly": anomaly,
    "anomaly_score": anomaly_score
}, index=X.index)

category_order = [
    "Distinction",
    "Pass",
    "Fail",
    "Withdrawn"
]

diagnostics["final_result"] = pd.Categorical(
    diagnostics["final_result"],
    categories=category_order,
    ordered=True
)

# 5. Calculate counts and percentages
summary = diagnostics.groupby(
    "final_result", observed=False
).agg(
    Total=("is_anomaly", "size"),
    Anomalies=("is_anomaly", "sum")
)

summary["Non_anomalies"] = (
    summary["Total"] - summary["Anomalies"]
)

summary["Anomaly_percentage"] = (
    100 * summary["Anomalies"] / summary["Total"]
)

print("\n========== OUTLIER SUMMARY BY FINAL RESULT ==========")
print(summary.round(3))

print("\n========== OVERALL RESULTS ==========")
print("Total learners:", len(diagnostics))
print("Detected anomalies:", int(anomaly.sum()))
print("Detected inliers:", int((~anomaly).sum()))
print("Observed anomaly percentage:",
      round(100 * anomaly.mean(), 3), "%")

# 6. Create Figure 4
fig, axes = plt.subplots(
    1, 3,
    figsize=(17, 5.5),
    gridspec_kw={"width_ratios": [1.2, 1, 1.3]}
)

# ------------------------------------------------------------
# (a) Anomaly and non-anomaly counts
# ------------------------------------------------------------
x_pos = np.arange(len(category_order))

anomaly_counts = summary.loc[
    category_order, "Anomalies"
].to_numpy()

normal_counts = summary.loc[
    category_order, "Non_anomalies"
].to_numpy()

axes[0].bar(
    x_pos, normal_counts,
    label="Non-anomalies"
)

axes[0].bar(
    x_pos, anomaly_counts,
    bottom=normal_counts,
    label="Anomalies"
)

axes[0].set_xticks(x_pos)
axes[0].set_xticklabels(category_order, rotation=25, ha="right")
axes[0].set_ylabel("Number of learners")
axes[0].set_title("(a) Outlier counts by final result")
axes[0].legend(frameon=False)

# ------------------------------------------------------------
# (b) Percentage of anomalies by final result
# ------------------------------------------------------------
percentages = summary.loc[
    category_order, "Anomaly_percentage"
].to_numpy()

bars = axes[1].bar(x_pos, percentages)

axes[1].set_xticks(x_pos)
axes[1].set_xticklabels(category_order, rotation=25, ha="right")
axes[1].set_ylabel("Anomalies (%)")
axes[1].set_title("(b) Outlier percentage by final result")

for bar, value in zip(bars, percentages):
    axes[1].text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f"{value:.2f}%",
        ha="center",
        va="bottom",
        fontsize=9
    )

# ------------------------------------------------------------
# (c) Anomaly-score distributions
# ------------------------------------------------------------
score_data = [
    diagnostics.loc[
        diagnostics["final_result"] == category,
        "anomaly_score"
    ].to_numpy()
    for category in category_order
]

axes[2].boxplot(
    score_data,
    tick_labels=category_order,
    showfliers=False
)

axes[2].set_ylabel("Anomaly score (-score_samples)")
axes[2].set_title("(c) Anomaly-score distributions")
axes[2].tick_params(axis="x", rotation=25)

# 7. Final layout
fig.suptitle(
    "Isolation Forest Outlier-Screening Diagnostics "
    "(contamination = 0.02)",
    fontsize=13,
    fontweight="bold"
)

plt.tight_layout()

# 8. Save publication-quality files
plt.savefig(
    "Fig4_IsolationForest.png",
    dpi=300,
    bbox_inches="tight"
)

plt.savefig(
    "Fig4_IsolationForest.pdf",
    bbox_inches="tight"
)

plt.show()

# 9. Save numerical results
summary.to_csv(
    "Fig4_Outlier_Summary.csv"
)

diagnostics.to_csv(
    "Fig4_Outlier_Diagnostics.csv",
    index=True
)

print("\nFiles saved:")
print("- Fig4_IsolationForest.png")
print("- Fig4_IsolationForest.pdf")
print("- Fig4_Outlier_Summary.csv")
print("- Fig4_Outlier_Diagnostics.csv")
```
